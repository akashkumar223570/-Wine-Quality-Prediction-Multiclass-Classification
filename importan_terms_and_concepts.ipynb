# 📊 Understanding Histograms

## What Is a Histogram?

A histogram is a graph that shows **how the values of a numerical feature are distributed**.

In simple words:

> It tells us where most of the values are located.

### Example

Suppose alcohol values are:

```
8.4
9.0
9.5
10.0
10.2
10.5
11.0
11.2
12.0
14.9
```

A histogram groups these values into ranges. For example:

```
8–9       → ██
9–10      → ███
10–11     → █████
11–12     → ███
12–13     → █
13–14     → 
14–15     → █
```

So we can quickly see: **where are most values concentrated?**

---

## What Does a Histogram Look Like?

Suppose we have the **Alcohol** feature. The histogram might look like:

```
Number
of wines
   ↑
500|             ███
400|           █████
300|         ███████
200|       █████████
100|    ███████████
   +--------------------→
      8  9  10  11  12  13  14  15
              Alcohol
```

- The **X-axis** shows the values.
- The **Y-axis** shows how many wines fall into those ranges.

---

## What Does `bins=20` Mean?

This is a very important concept.

```python
features.hist(bins=20)
```

`bins` means: **how many groups should we divide the values into?**

### Example

Suppose alcohol ranges from:

```
8 → 15
```

- If we use `bins=7`, we create roughly **7 intervals**.
- If we use `bins=20`, we create **20 intervals**.

Think of it like putting values into boxes 📦:

```
Bin 1
Bin 2
Bin 3
Bin 4
...
Bin 20
```

Each bar represents one interval.

### Simple Meaning

> `bins=20` = divide the numerical range into 20 groups and count how many values fall into each group.

# 🔗 Understanding Correlation and Heatmaps

## 1. What Is Correlation?

This is the most important concept before understanding the graph.

> **Correlation** means how strongly two numerical variables move together.

### Positive Correlation

For example, suppose:

```
Alcohol ↑
Quality ↑
```

If quality tends to increase when alcohol increases, they have a **positive correlation**.

### Negative Correlation

On the other hand:

```
Volatile acidity ↑
Quality ↓
```

If quality tends to decrease when volatile acidity increases, they have a **negative correlation**.

### No Correlation

If there is no clear relationship:

```
Feature A ↑/↓
Quality doesn't follow a clear pattern
```

then the correlation is close to **0**.

---

## 2. Correlation Has a Range

Correlation values fall between **-1** and **+1**:

```
-1       0       +1
│────────│────────│
Negative  None   Positive
```

- **+1** — Very strong positive relationship. One increases → the other tends to increase.
- **0** — No linear relationship. One changes → the other doesn't show a clear linear pattern.
- **-1** — Very strong negative relationship. One increases → the other tends to decrease.

---

## 3. Easy Way to Remember

| Correlation | Meaning |
|-------------|---------|
| +1 | Strong positive relationship |
| +0.8 | Strong positive |
| +0.5 | Moderate positive |
| +0.2 | Weak positive |
| 0 | No linear relationship |
| -0.2 | Weak negative |
| -0.5 | Moderate negative |
| -0.8 | Strong negative |
| -1 | Strong negative |

**Summary:**

- Closer to **+1** → strong positive relationship
- Closer to **-1** → strong negative relationship
- Closer to **0** → weak/no linear relationship

---

## 4. What Is a Heatmap?

A heatmap uses colors to represent numerical values.

> A heatmap is a visual data tool that uses colors to show values, numbers, or activity levels across a two-dimensional grid or map.

```
Number
  ↓
Color
  ↓
Easy visual understanding
```

Instead of reading raw numbers such as:

```
0.52
-0.41
0.13
0.68
```

we can quickly look at the colors and identify stronger/weaker relationships at a glance.

# 📊 Understanding `countplot()`

## 1. What Is `countplot()`?

`countplot()` is a **Seaborn** function used to count how many times each category/value appears and display the counts as bars.

**Basic syntax:**

```python
sns.countplot(data=df, x="column_name")
```

**For our project:**

```python
sns.countplot(
    data=df,
    x="quality"
)
```

This means: take the `quality` column, count how many wines have each quality value, and show those counts as bars.

---

## 2. When Do We Use `countplot()`?

Use `countplot()` when you want to answer:

> "How many observations belong to each category?"

### Common Examples

**Gender:**
```
Male      → 500
Female    → 450
```

**Department:**
```
IT         → 300
HR         → 150
Finance    → 200
```

**Quality:**
```
3 → 10 wines
4 → 53 wines
5 → 577 wines
6 → 535 wines
7 → 167 wines
8 → 17 wines
```

So `countplot()` is particularly useful for **categorical data**.

---

## 3. Why Is It Important in Machine Learning?

Before training a classification model, we should check:

> "Are our classes balanced or imbalanced?"

For example:

| Class | Number |
|-------|--------|
| Low | 63 |
| Medium | 1112 |
| High | 184 |

This is clearly **imbalanced**. The model has lots of examples for Medium, but few for Low/High — and this can affect model performance.

So a countplot is often used during **EDA (Exploratory Data Analysis)**.

---

## 4. Basic `countplot()` Syntax

```python
sns.countplot(
    data=df,
    x="quality"
)
```

There are two important parameters here:

- **`data`** — `data=df` means: use the DataFrame `df`.
- **`x`** — `x="quality"` means: put the `quality` column on the X-axis.

---

## 5. Most Important Parameters

| Parameter | What It Does | Importance |
|-----------|----------------|------------|
| `data` | Specifies the DataFrame | ⭐⭐⭐⭐⭐ |
| `x` | Category on X-axis | ⭐⭐⭐⭐⭐ |
| `y` | Category/value on Y-axis | ⭐⭐⭐⭐ |
| `hue` | Creates groups using another column | ⭐⭐⭐⭐⭐ |
| `order` | Controls category order | ⭐⭐⭐⭐ |
| `hue_order` | Controls order of hue categories | ⭐⭐⭐ |
| `palette` | Controls colors | ⭐⭐⭐ |
| `color` | Sets one color for all bars | ⭐⭐ |
| `ax` | Specifies which Matplotlib axes to draw on | ⭐⭐ |
| `stat` | Controls what is shown instead of raw count (in supported Seaborn versions) | ⭐⭐ |

For this project, focus mainly on: **`data`, `x`, `hue`, `order`, `palette`**.

---

## 6. `data`

```python
sns.countplot(data=df, x="quality")
```

`data=df` tells Seaborn: *"The column I'm talking about is inside the DataFrame called `df`."*

Without `data`, you can also provide a Series directly:

```python
sns.countplot(x=df["quality"])
```

But when working with DataFrames, this is usually cleaner:

```python
sns.countplot(data=df, x="quality")
```

---

## 7. `x`

```python
x="quality"
```

This tells Seaborn to put `quality` on the X-axis. For example:

```python
sns.countplot(data=df, x="quality")
```

 
---

## 8. `y`

You can also make a **horizontal** countplot:

```python
sns.countplot(
    data=df,
    y="quality"
)
```

Instead of:

```
       █
       █
       █
───────5──────→
```

you get something like:

```
Quality
   5 █████████████
   6 ████████████
   7 ████
   8 █
       → Number of wines
```

### When Is `y` Useful?

When category names are long, for example:

```
Computer Science
Mechanical Engineering
Electrical Engineering
Information Technology
```

A horizontal chart is much easier to read.

---

## 9. `hue` ⭐⭐⭐⭐⭐

This is one of the most useful parameters.

`hue` means: use another column to divide the bars into groups.

Suppose your dataset has `quality` and `quality_category`. You could write:

```python
sns.countplot(
    data=df,
    x="quality",
    hue="quality_category"
)
```

Now Seaborn uses `quality_category` to group the observations. Conceptually:

```
Quality 5 → Medium
Quality 6 → Medium
Quality 7 → High
```

`hue` becomes very useful when you want to compare categories inside another category.

---

## 10. Another Example of `hue`

Suppose we have:

| Gender | Quality |
|--------|---------|
| Male | 5 |
| Female | 6 |
| Male | 6 |
| Female | 5 |

We can write:

```python
sns.countplot(
    data=df,
    x="quality",
    hue="gender"
)
```

Now for each quality score, we can see separate bars for Male and Female. So instead of asking:

> "How many wines have quality 5?"

we can ask:

> "How many quality-5 wines belong to each group?"

---

## 11. `order`

Suppose your categories are: `Medium, Low, High`, but you want: `Low, Medium, High`. You can specify:

```python
sns.countplot(
    data=df,
    x="quality_category",
    order=["Low", "Medium", "High"]
)
```

`order` means: tell Seaborn exactly what order to display the categories in. This is very useful for ordered categories.

---

## 12. `hue_order`

If you use `hue`, you can also control the order of hue categories:

```python
sns.countplot(
    data=df,
    x="quality",
    hue="quality_category",
    hue_order=["Low", "Medium", "High"]
)
```

So:
- `order` → order of X/Y categories
- `hue_order` → order of hue groups

---

## 13. `palette`

You can control the colors using:

```python
palette="Set2"
```

Example:

```python
sns.countplot(
    data=df,
    x="quality_category",
    order=["Low", "Medium", "High"],
    palette="Set2"
)
```

This is mainly for visual presentation — it doesn't change the underlying data or ML model.

> **Important distinction:** `palette` controls **appearance**, not data processing.

---

## 14. `color`

If you want all bars to have one color:

```python
sns.countplot(
    data=df,
    x="quality",
    color="skyblue"
)
```

**Difference:**
- `color` → one color for all bars
- `palette` → can provide different colors for categories/groups

---

## 15. `stat`

In current Seaborn versions, `countplot()` supports the `stat` parameter.

By default, `stat="count"` means: show the actual number of observations (e.g., Quality 5 → 577 wines).

You can also use other supported statistics such as proportions/percentages, depending on your Seaborn version:

```python
sns.countplot(
    data=df,
    x="quality",
    stat="percent"
)
```

Now the Y-axis represents percentage of observations rather than the raw number.

**For your beginner stage, remember:**
- `stat="count"` → actual number
- `stat="percent"` → percentage

---

## 16. `ax`

This is a little more advanced. If you have multiple plots in one figure, you can tell Seaborn where to draw the countplot:

```python
fig, ax = plt.subplots()

sns.countplot(
    data=df,
    x="quality",
    ax=ax
)
```

You don't need to worry about this yet — just remember that `ax` tells Seaborn which plotting area to use.

## 18. Countplot vs. Histogram

This distinction is very important, since you already learned about histograms.

### Countplot

Used mainly for **categories**:

```python
sns.countplot(data=df, x="quality")
```

`Quality: 3, 4, 5, 6, 7, 8`

### Histogram

Used for the **distribution of numerical values**:

```python
sns.histplot(data=df, x="alcohol")
```

`Alcohol: 8.4, 8.7, 9.1, 9.5, 10.2, 11.4, ...`

### Simple Comparison

| | Countplot | Histogram |
|---|-----------|-----------|
| Main purpose | Count categories | Show numerical distribution |
| Example | Quality | Alcohol |
| Bars represent | Categories | Numerical ranges/bins |
| Typical function | `sns.countplot()` | `sns.histplot()` |

**Easy memory:**
- Countplot → "How many of each category?"
- Histogram → "How are numerical values distributed?"

---

## 19. Countplot vs. Barplot

Another common point of confusion.

### `countplot()`

You give it the data and it counts automatically:

```python
sns.countplot(data=df, x="quality")
```

You don't calculate the counts yourself.

### `barplot()`

Usually used when you already have a numeric measurement/statistic to compare across categories. For example:

```
Quality 5 → average alcohol = 9.8
Quality 6 → average alcohol = 10.2
Quality 7 → average alcohol = 11.1
```

A barplot could show those averages.

**Remember:**
- `countplot` → COUNT
- `barplot` → MEASURE/STATISTIC
