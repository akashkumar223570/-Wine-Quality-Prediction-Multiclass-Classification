# 🍷 Wine Quality — Feature Reference Guide

## Attribute Overview

| # | Attribute | What Does It Mean? | Why Do We Use It? | If LOW | If HIGH |
|---|-----------|---------------------|---------------------|--------|---------|
| 1 | **Fixed Acidity** | Acids in wine that do not easily evaporate. | Helps determine the freshness, sharpness, and taste balance of wine. | Wine may taste flat or less fresh. | Wine may taste very sour or sharp. |
| 2 | **Volatile Acidity** | Acidity that can evaporate easily, mainly related to acetic acid. | Helps identify vinegar-like taste or smell. | Usually better; less vinegar-like character. | Can give a vinegar-like smell/taste and may reduce quality. |
| 3 | **Citric Acid** | An acid that can give wine a fresh and citrus-like character. | Helps describe the freshness and flavor of wine. | Less freshness. | More fresh/citrus character; too much can make wine too acidic. |
| 4 | **Residual Sugar** | Sugar that is left after fermentation. | Helps determine how sweet or dry the wine is. | Wine is usually drier and less sweet. | Wine is sweeter. |
| 5 | **Chlorides** | The amount of chloride-related compounds in the wine. | Helps describe the wine's salt/mineral-related character. | Less salty character. | More salty character; too much can negatively affect taste. |
| 6 | **Free Sulfur Dioxide** | SO₂ that is still free and available in the wine. | Helps protect wine from oxidation and some microorganisms. | Less protection; wine may be more vulnerable to oxidation. | More protection, but too much may cause unwanted sulfur-like smell/taste. |
| 7 | **Total Sulfur Dioxide** | Free SO₂ + Bound SO₂. It is the total amount of SO₂. | Shows the total sulfur dioxide present in the wine. | Less overall SO₂ protection. | More SO₂; very high amounts can be undesirable. |
| 8 | **Density** | How heavy a liquid is for its volume. | Gives information about the wine's sugar, alcohol, and overall composition. | Can be associated with higher alcohol/lower sugar, depending on the wine. | Can be associated with more sugar/dissolved material, depending on the wine. |
| 9 | **pH** | A measure of how acidic the wine is. | Important for taste, stability, and microbial control. | More acidic wine. | Less acidic wine. |
| 10 | **Sulphates** | Sulfate-related compounds present in the wine. | Gives information about the wine's chemical composition and preservation-related properties. | Less sulfate-related content. | More sulfate-related content; very high levels can be undesirable. |
| 11 | **Alcohol** | The amount of alcohol in wine, usually shown as % by volume. | Affects the wine's strength, body, and warmth. | Wine may feel lighter and less strong. | Wine may feel stronger, fuller, and warmer; too much can feel harsh/heavy. |

## 🧠 Super-Simple Memory Trick

| Feature | Think About |
|---------|-------------|
| Fixed Acidity | 🍋 Basic acidity |
| Volatile Acidity | 🥫 Vinegar |
| Citric Acid | 🍋 Fresh/citrus |
| Residual Sugar | 🍬 Sweetness |
| Chlorides | 🧂 Salt |
| Free SO₂ | 🛡️ Available protection |
| Total SO₂ | 🛡️ Total protection chemical |
| Density | ⚖️ Heaviness |
| pH | 🧪 Acidity level |
| Sulphates | 🧪 Sulfur-related compound |
| Alcohol | 🍷 Strength |

## ⚠️ Important Note for Your ML Project

> Don't assume **HIGH = good** or **LOW = bad**. The quality of wine depends on the **combination of all 11 features**, not one feature alone.
