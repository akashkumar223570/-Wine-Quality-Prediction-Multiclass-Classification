# Confusion Matrix Analysis — Random Forest vs SGD vs SVC

## 1. First Understand How to Read a Confusion Matrix

In all three matrices:

```
Rows    = Actual class
Columns = Predicted class
```

So:

|  | Predicted Low | Predicted Medium | Predicted High |
|---|---|---|---|
| **Actual Low** | | | |
| **Actual Medium** | | | |
| **Actual High** | | | |

### Most Important Rule ⭐

> **Diagonal values = Correct predictions**

The diagonal is:

```
Low    → Low
Medium → Medium
High   → High
```

Everything outside the diagonal = **Wrong prediction**.

## 9. But Here Is the VERY Important Point 🚨

Don't conclude:

> "SVC is definitely the best model because it has 84.6% accuracy."

**Why?**

Look at the Low class.

```
SVC:
Actual Low = 13
Correct    = 0
```

SVC completely fails to identify Low wines.

**Why can it still get 84.6% accuracy?**

Because there are many more Medium wines.

Remember:

```
Low     = 13
Medium  = 222
High    = 37
```

The dataset is **imbalanced**.

So getting most Medium wines correct gives a large contribution to overall accuracy.

## 11. Why Is This Happening?

The main reason is likely the **class imbalance**.

Your test set contains:

```
Medium → 222
High   → 37
Low    → 13
```

Think about it:

```
Medium
████████████████████████████████████

High
██████

Low
██
```

The model has far more Medium examples to learn from.

So it becomes much easier for the model to learn:

> "When I'm uncertain, Medium is a common answer."

This is why many Low and High wines are being predicted as Medium.

## 12. How to Understand the Color

Your heatmap uses colors.

You don't need to focus too much on color.

> The number written inside each box is the most important thing.

For example:

|  | Predicted Low | Predicted Medium | Predicted High |
|---|---|---|---|
| **Actual Low** | 2 | 11 | 0 |

The **209** in Random Forest:

```
Actual Medium
Predicted Medium
      ↓
     209
```

is very bright because it is a large number.

So:

> Color mainly helps us visually notice large and small numbers. The actual number is what we use for interpretation.

---

## 13. Easy Way to Remember Confusion Matrix

Always ask three questions:

**Question 1** — What is the actual class?
> Look at the row.

**Question 2** — What did the model predict?
> Look at the column.

**Question 3** — Is it on the diagonal?
> YES → Correct ✅
> NO → Wrong ❌

## 14. Your Final Interpretation

You can write this in your project report:

> The confusion matrices show that all three models classify Medium-quality wines very well, but they struggle to identify Low- and High-quality wines. Most errors occur when Low or High wines are incorrectly classified as Medium. SVC achieves the highest overall accuracy (about 84.6%) and correctly classifies the most Medium wines, while SGD correctly identifies the most High wines. Random Forest performs between the two in overall accuracy. However, none of the models performs well on the Low class, mainly because the dataset contains far fewer Low samples than Medium samples. Therefore, accuracy alone should not be used to select the final model; precision, recall, and F1-score for each class should also be considered.
