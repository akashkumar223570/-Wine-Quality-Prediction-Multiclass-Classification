# 🍷 Wine Quality Prediction — Problem Understanding

## 1. What Problem Are We Solving?

We have a dataset containing information about different wines. For every wine, we know some of its chemical/physicochemical properties, such as:

- Acidity
- Sugar
- Chlorides
- Sulfur dioxide
- Density
- pH
- Sulphates
- Alcohol

We also have a **quality score** for each wine. For example:

| Alcohol | pH   | Density | Sulphates | Quality |
|---------|------|---------|-----------|---------|
| 9.4     | 3.51 | 0.9978  | 0.56      | 5       |
| 9.8     | 3.20 | 0.9968  | 0.68      | 5       |
| 10.5    | 3.26 | 0.9970  | 0.65      | 6       |
| 12.0    | 3.30 | 0.9950  | 0.75      | 7       |

The question is:

> **Can we use the chemical properties of a wine to predict its quality?**

That is the problem we aim to solve.

---

## 2. What Are We Giving the Machine?

We give the machine the chemical properties of the wine. For example:

- Fixed acidity = 7.4
- Volatile acidity = 0.70
- Citric acid = 0.00
- Residual sugar = 1.9
- Chlorides = 0.076
- ...
- Alcohol = 9.4

These are our **inputs**. In machine-learning terminology, we call them **Features**.

```
Chemical properties
        ↓
     FEATURES
```

---

## 3. What Do We Want the Machine to Predict?

We want the machine to predict the **Wine Quality**. This is our **target variable**.

Initially, the dataset contains quality scores such as:

```
3, 4, 5, 6, 7, 8
```

So the basic problem looks like:

```
Features
   ↓
Machine Learning Model
   ↓
Quality
```

---

## 4. Why Are We Changing Quality into Low / Medium / High?

This is an important part of our project.

The original quality values are:

```
3
4
5
6
7
8
```

However, the number of wines in each category is **not equal**. For example, there are many wines with quality 5 and 6, but relatively few wines with quality 3 or 8.

This is called **class imbalance**.

This creates a problem. Imagine our model sees:

- Quality 5 → 680 wines
- Quality 6 → 638 wines
- Quality 3 → very few
- Quality 8 → very few

The model gets many opportunities to learn about classes 5 and 6, but very few opportunities to learn about classes 3 and 8.

---

## 5. Our Solution — Group the Quality Scores

Instead of asking the model to predict:

```
3 / 4 / 5 / 6 / 7 / 8
```

we create three categories:

- **3, 4 → LOW**
- **5, 6 → MEDIUM**
- **7, 8 → HIGH**

So our new problem becomes:

> **Given the chemical properties of a wine, predict whether its quality is Low, Medium, or High.**

This is the classification problem we will solve.

---

## 6 What Are the 3 Models Doing?

We are going to give the same problem to three different algorithms.

### Model 1 — Random Forest
> "Let's use many decision trees and combine their decisions."

### Model 2 — SGD Classifier
> "Let's gradually learn the mathematical decision boundary by minimizing errors."

### Model 3 — SVC
> "Let's find boundaries that separate the different classes."

All three receive essentially the same input:

```
Wine chemical properties
```

and produce:

```
Low / Medium / High
```

Then we compare them.

---

## 7. Why Are We Using Three Models?

Suppose we use only Random Forest. We might get:

```
Random Forest → 82%
```

But we don't know whether another algorithm could perform better. So we try:

```
                Same Dataset
                     ↓
        ┌────────────┼────────────┐
        ↓            ↓            ↓
 Random Forest      SGD          SVC
        ↓            ↓            ↓
       82%          75%          80%
```

Now we can compare them. This is the main reason for using multiple classifiers.

---

## 7. How Do We Decide Which Model Is Best?

We cannot simply look at one prediction. We will test the models using **unseen test data** and calculate the following metrics:

- **Accuracy** — How many predictions were correct overall?
- **Precision** — When the model predicts a class, how often is it correct?
- **Recall** — How many of the actual examples of a class did the model correctly identify?
- **F1-Score** — A combined measure of precision and recall.

And we will use a **Confusion Matrix**, which shows which classes the model is confusing with each other. For example:

- Actual Medium → Predicted High
- Actual High → Predicted Medium

---

## 8. So What Exactly Is the Final Goal?

Our final goal is not merely:

> "Predict wine quality."

It is more specifically:

> **Build a multiclass classification system that predicts Low, Medium, or High wine quality from physicochemical properties, train three different classifiers, evaluate their performance, and identify the most suitable model.**
