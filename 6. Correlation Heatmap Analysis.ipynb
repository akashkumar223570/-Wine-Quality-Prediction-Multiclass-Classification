| Correlation | Simple Meaning |

- **Negative (-)** → they tend to move in opposite directions.

## 3. ⭐ Alcohol → Quality = +0.48

This is the **strongest positive correlation** with quality among these features.

```
Alcohol ↑
    ↓
Quality tends to ↑
```

**+0.48 means:** In this dataset, wines with higher alcohol tend to have higher quality scores.

> ❌ It does **NOT** mean "higher alcohol always means better wine." It only tells us there is a moderate positive linear relationship in this dataset.

## 4. ⭐ Volatile Acidity → Quality = -0.40

This is the **strongest negative relationship** with quality among these features.

```
Volatile acidity ↑
          ↓
Quality tends to ↓
```

**-0.40 means:** Wines with higher volatile acidity tend to have lower quality scores.

This makes sense from a wine perspective, because excessive volatile acidity can produce undesirable vinegar-like characteristics.

> ❌ Don't say: "If volatile acidity increases, quality must decrease."
> ✅ Instead say: "Higher volatile acidity is moderately associated with lower quality in this dataset."

## 14. What About Relationships Between the Features?

This heatmap also tells us how the input features relate to each other. There are some interesting ones.

### Fixed Acidity ↔ Citric Acid: +0.67

A relatively strong positive relationship. Wines with higher fixed acidity tend to also have higher citric acid in this dataset.

### Fixed Acidity ↔ pH: -0.69

A relatively strong negative relationship. Higher fixed acidity tends to be associated with lower pH.

> Remember: lower pH = more acidic. So this relationship makes chemical sense.

### Free SO₂ ↔ Total SO₂: +0.67

Also a strong positive relationship. This is logical because:

```
Total SO₂ = Free SO₂ + Bound SO₂
```

So when free SO₂ increases, total SO₂ often tends to increase as well.

### Density ↔ Alcohol: -0.50

A moderate negative relationship. Higher alcohol tends to be associated with lower density in this dataset.

## 🎯 What Does This Mean for Our ML Project?

Some features have noticeable relationships with quality:

```
Alcohol          +0.48
Volatile acidity -0.40
```

while several others have very weak linear relationships:

```
Residual sugar       +0.01
Free SO₂             -0.05
pH                   -0.06
```

**But do not remove the weak-correlation features yet.**

**Why?** Correlation measures only a **linear** relationship between two variables. A machine-learning model can find more complicated relationships and interactions between multiple features. For example:

```
Alcohol
   +
Sulphates
   +
Acidity
   +
Density
   +
Other features
       ↓
    ML Model
       ↓
   Quality
```

So we keep the features unless we have a good reason to remove them.
